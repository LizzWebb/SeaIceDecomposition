```python
import pylab as plt
from xmitgcm import open_mdsdataset
import numpy as np
%matplotlib ipympl
```

```python
data=open_mdsdataset('./',prefix=['SIsig1','SIsig2','SIarea','SIheff'],read_grid=False)
```

```python
data
```

```python
sI=0.5*(data.SIsig1+data.SIsig2)
sII=0.5*(data.SIsig1-data.SIsig2)

sI_ell=np.arange(-1,0,0.0001)
sII_ell=0.5*np.sqrt(-sI_ell*(sI_ell+1))

fig,ax=plt.subplots()
p=ax.plot(sI[-1],sII[-1],'r.',ms=0.5)
p2=ax.plot(sI_ell,sII_ell,'k-')
ax.set_aspect('equal')
```
